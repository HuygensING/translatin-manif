# Step by step

Below you can inspect all the steps of the conversion:

```python
%load_ext autoreload
%autoreload 2
```

```python
from tf.app import use
from tf.convert.pagexml import PageXML
from tf.convert.watm import WATM, WATMS
from tf.advanced.helpers import dm
```

```python
BACKEND = "gitlab.huc.knaw.nl"
ORG = "translatin"
REPO = "corpus"

REPO_DIR = f"~/{BACKEND}/{ORG}/{REPO}"
SOURCE_DIR = f"{REPO_DIR}/organized/source"
```

# Step 1: Convert

```python
P = PageXML(SOURCE_DIR, REPO_DIR, verbose=-1, source=0, tf="0.1.2")
```

```python
P.task(convert=True, verbose=-1)
```

# Step 2: Load

```python
P.task(load=True, verbose=-1)
```

# Step 3: Compile app

```python
P.task(app=True, verbose=1)
```

# Step 4: use

```python
A = use(f"{ORG}/{REPO}:clone", relative="tf/M95", backend=BACKEND, checkout="clone", hoist=globals())
```

# Step 5: Inspect

```python
def sample(doc, page, startLine, endLine):
    for (label, fmt) in ("physical", "text-orig-full"), ("logical", "text-logic-full"):
        A.dm(f"### {label}\n")
        for ln in range(startLine, endLine + 1):
            lNode = A.nodeFromSectionStr(f"{doc}@{page}:{ln}")
            A.plain(lNode, fmt=fmt)
    A.dm("### Both\n")
    for ln in range(startLine, endLine + 1):
        lNode = A.nodeFromSectionStr(f"{doc}@{page}:{ln}")
        A.pretty(lNode, withNodes=True, extraFeatures="str")
```

```python
sample("M95", 2, 2, 6)
```

# Step 6: Generate WATM

```python
W = WATMS(ORG, REPO, BACKEND, "pagexml", skipMeta=False)
```

```python
# W.produce(doc="M38")
W.produce()
```

# Step 7: Test

We test only a single document.

```python
A = use(f"{ORG}/{REPO}:clone", relative="tf/M95", checkout="clone", backend=BACKEND, hoist=globals())
```

```python
skipMeta = False
WA = WATM(A, "pagexml", skipMeta=skipMeta)
WA.makeText()
WA.makeAnno()
WA.writeAll()
```

```python
WA.testAll()
```
