```python
import re

tokenRe = re.compile(r"""\w+|\W""")


def tokenize(line):
    tokens = []

    for word in line.split():
        ts = [[t, ""] for t in tokenRe.findall(word)]
        if len(ts):
            ts[-1][-1] = " "
        tokens.extend(ts)

    if len(tokens):
        tokens[-1][-1] = "\n"
    return tuple(tokens)


def repTokens(tokens):
    text = []
    for (t, space) in tokens:
        text.append(f"‹{t}›{space}")
    return "".join(text)


def unwrap(x, indent=0):
    tp = x.main_type
    xid = x.id
    box = x.coords.box
    pre = " " * indent
    print(f"{pre}- {tp} {xid} in {box}")

    iindent = indent + 2
    
    if tp == "scan":
        for y in x.get_text_regions_in_reading_order():
            unwrap(y, indent=iindent)
        for y in x.lines:
            unwrap(y, indent=iindent)
    elif tp == "text_region":
        for y in x.get_text_regions_in_reading_order():
            unwrap(y, indent=iindent)
        for y in x.lines:
            unwrap(y, indent=iindent)
    elif tp == "line":
        print(f"{pre}  {repTokens(tokenize(x.text))}")
    else:
        print(f"{pre}  UNKNOWN TYPE")
        
```
